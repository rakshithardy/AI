```python
from itertools import product

def entails():

    n = int(input("Enter number of propositions: "))
    symbols = input("Enter propositions: ").lower().split()

    if len(symbols) != n or len(set(symbols)) != n:
        print("Invalid propositions!")
        return

    m = int(input("Enter number of KB statements: "))

    kb = []
    for i in range(m):
        kb.append(input(f"Enter KB {i+1}: ").lower())

    q = int(input("Enter number of queries: "))

    queries = []
    for i in range(q):
        queries.append(input(f"Enter Query {i+1}: ").lower())

    def evaluate(expr, model):
        return eval(expr, {"__builtins__": {}}, model)

    models = list(product([True, False], repeat=n))

    headers = (
        [s.upper() for s in symbols]
        + [f"KB{i+1}" for i in range(m)]
        + ["KB"]
        + [f"Q{i+1}" for i in range(q)]
    )

    print("\n========== TRUTH TABLE ==========\n")

    print(" | ".join(f"{h:^5}" for h in headers))
    print("-" * (len(headers) * 8))

    results = [True] * q

    for values in models:

        model = dict(zip(symbols, values))

        kb_values = [evaluate(expr, model) for expr in kb]
        kb_true = all(kb_values)

        query_values = [evaluate(expr, model) for expr in queries]

        row = list(values) + kb_values + [kb_true] + query_values

        print(" | ".join(f"{'T' if x else 'F':^5}" for x in row))

        for i in range(q):
            if kb_true and not query_values[i]:
                results[i] = False

    print("\n========== ENTAILMENT RESULTS ==========")

    for i in range(q):
        print(
            f"Query {i+1}:",
            "Entailed" if results[i] else "Not Entailed"
        )


entails()
```
