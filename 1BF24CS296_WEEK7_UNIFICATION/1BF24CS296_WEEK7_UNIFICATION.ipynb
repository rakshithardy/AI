```python

def is_var(x):
    return isinstance(x, str) and x.startswith("?")

def unify(x, y, s):
    if x == y:
        return s

    if is_var(x):
        if x in y:
            return None
        s[x] = y
        return s

    if is_var(y):
        if y in x:
            return None
        s[y] = x
        return s

    if len(x) != len(y) or x[0] != y[0]:
        return None

    for a, b in zip(x[1:], y[1:]):
        s = unify(a, b, s)
        if s is None:
            return None

    return s


# Main program
print("Enter first expression:")
E1 = tuple(input().split())

print("Enter second expression:")
E2 = tuple(input().split())

result = unify(E1, E2, {})

if result is None:
    print("UNIFICATION FAILURE")
else:
    print("UNIFICATION SUCCESS")
    print("Substitution:", result)
```
